```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
np.random.seed(2)
n=35
size1=1000000
mean_p=100
std_p=15
x_p=np.random.normal(mean_p,std_p,size1)
poe1=np.mean(np.random.choice(x_p,35,replace=False))
poe2=np.mean(np.random.choice(x_p,35,replace=False))
poe3=np.mean(np.random.choice(x_p,35,replace=False))

plt.figure(figsize=(12,6))
sns.kdeplot(x_p)
plt.plot(poe1,0.005,marker="*",ms=40,mfc="red",mec="black",label="Point of Estimate 1",lw=0.3)
plt.axvline(poe1,color="red")
plt.plot(poe2,0.0125,marker="o",ms=40,mfc="cyan",mec="black",label="Point of Estimate 1",lw=0.2)
plt.axvline(poe2,color="cyan")
plt.plot(poe3,0.020,marker="v",ms=40,mfc="yellow",mec="black",label="Point of Estimate 1",lw=0.2)
plt.axvline(poe3,color="yellow")
plt.axvline(mean_p,lw=20,alpha=0.2)
plt.legend(facecolor="beige",ncol=3,loc=(0,1.15))
plt.title("Point of Estimate ",x=0.26,y=1.3,size=45,color="purple")
plt.ylabel("Probability",color="green")
plt.xlabel("IQ Score",color="red")
plt.annotate("Point of estimate\nsample mean vs population mean",
             size=18,
             color="blue",
             xytext=(115,0.02),
             bbox=dict(boxstyle="sawtooth,pad=0.5",facecolor="yellow",edgecolor="red"),
             xy=(100,0.0075,),
             arrowprops=dict(arrowstyle="fancy",facecolor="red",edgecolor="indigo",connectionstyle="arc3,rad=0.25"))
plt.show()
```
